### 머신러닝이란?
- 기계가 스스로 학습하고 예측하는 기술

### 지도학습 VS 비지도학습

  - 지도학습 : 문제와 답을 쌍으로 넣어서 학습하는 방법
      - 학습성공률이 높다, 그러나 정답을 만들기 위한 자원(시간, 돈 등)이 많이 들어간다
  - 비지도학습 : 정답이 없는 데이터의 패턴을 찾는 학습 방법

### 머신러닝 프로세스
1. 문제정의
2. 데이터 수집
3. 탐색적 데이터 분석(EDA) & 전처리
4. 모델선택
  - KNN
  - DEcision Tree
  - Linear model
5. 하이퍼파라미터 튜닝
  - KNN : 이웃의 숫자
  - Decision Tree : tree의깊이
  - Linear model : 규제 정도
6. 모델평가
  - 분류 : 모델의 예측 값이 정해진 선택지에서 나오는 경우 (ex :혈액형 맞추기, 성별 맞추기 등)
    - 정확도, 재현율, 정밀도 등
  - 회귀 : 지정된 범위 없는 수치데이터를 맞추는 경우
  (ex : 보통 실수형태로 추론하는 경우가 많다)
    - R2, MSE 등
